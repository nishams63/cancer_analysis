# Rare Mutation Combination Discovery
Identifies co-occurring genomic alterations occurring at < 0.5% frequency in the cohort.


### 1. Load Mutation Co-occurrence Matrix


```python
import pandas as pd
df_cooc = pd.read_parquet('../../data/processed/mutation_cooccurrence.parquet')
df_cooc.head(10)
```

### 2. Mine Candidate Rare Combinations


```python
from stage5.src.eda.rarity_analysis import MutationRarityAnalyzer
df_mut = pd.read_parquet('../../data/processed/mutation_frequencies.parquet')
rarity = MutationRarityAnalyzer()
df_rare = rarity.analyze(df_cooc, df_mut)
print(f'Identified {len(df_rare[df_rare["candidate_stress_test"]])} candidate rare pairs.')
df_rare[df_rare['candidate_stress_test']].head()
```
